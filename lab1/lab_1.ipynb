# Ćwiczenie 1

```python
import numpy as np

tablica = np.arange(10, 51)
print(tablica)

# liczba wymiarów
print("liczba wymiarow:", tablica.ndim)
# rozmiar tablicy
print("rozmiar:", tablica.shape)
# liczba elementów
print("liczba elem:", tablica.size)
```

```python
# suma elementów
print("suma:", np.sum(tablica))
# średnia
print("srednia:", np.mean(tablica))
# wartośc maksymalna
print("max:", np.max(tablica))
# wartość minimalna
print("min:", np.min(tablica))
# mediana
print("mediana:", np.median(tablica))
# odchylenie standardowe
print("odchyl std:", np.std(tablica))
```

## Utworzenie nowej tablicy zawierającej 20 ostatnich elementów i elementy o nieparzystych indeksach.

```python
nowa_tablica = tablica[-20:]
nowa_tablica = tablica[1::2]
nowa_tablica
```

## Wygenerowanie tablicy 50 liczb losowych z zakresu od 1 do 100

```python
tablica_losowe = np.random.randint(1, 101, 50)
tablica_losowe
```

## Posortowanie wygenerowanej tablicy rosnąco i malejąco.

```python
tablica_malejaco = np.sort(tablica_losowe)[::-1]
print(tablica_malejaco)

tablica_rosnąco = np.sort(tablica_losowe)
print(tablica_rosnąco)
```

## Obliczenie mediany i odchylenia standardowego wygenerowanej tablicy.

```python
print("mediana:", np.median(tablica_losowe))
print("odchylenie std:", np.std(tablica_losowe))
```

# Ćwiczenie 2

```python
import pandas as pd
import matplotlib.pyplot as plt
fuel = pd.read_csv("../pliki/fuel.csv")
fuel
```

## wyświtlenie pięciu pierwszych wierszy

```python
fuel.head()
```

# Wyświetlenie liczby wierszy i kolumn oraz nazwy wszystkich kolumn

```python
print(fuel.shape)
print("Liczba wierszy:", fuel.shape[0])
print(print("Liczba kolumn:", fuel.shape[1]))
list(fuel.columns)
```

# Wyświetlenie podstawowe statystyki opisowe dla danych liczbowych.

```python
fuel.describe()
```

# Sprawdzenie, czy zbiór danych zawiera brakujące wartości. Jeśli tak, wyświetlić liczbę brakujących wartości dla każdej kolumny

```python
print(fuel.isnull().any())
print(fuel.isnull().sum())
# NIE ZAWIERA
```

# Wykonanie wykresu liniowego przedstawiającego łączne zużycie paliwa dla pierwszych 20 samochodów

```python
plt.figure(figsize=(8,5))
plt.plot(fuel["FUELCONSUMPTION_CITY"].head(20),
 marker='o',
 linestyle='-',
 label="zużycie paliwa")
plt.title("zużycie paliwa dla pierwszych 20 samochodów")
plt.xlabel("Numer samochodu")
plt.ylabel("zużycie paliwa")
plt.legend()
plt.savefig("wykres_liniowy.png")
plt.show()
```

# Wykonanie wykres punktowy przedstawiający zależność pomiędzy pojemnością silnika a emisją CO2.

```python
plt.figure(figsize=(8,5))
plt.scatter(fuel["ENGINESIZE"], fuel["CO2EMISSIONS"],
label="emisja")
plt.title("Pojemnosc silnika a emisja CO2")
plt.xlabel("Pojemnosc silnika")
plt.ylabel("emisja CO2")
plt.legend()
plt.savefig("wykres_punktowy.png")
plt.show()
```

# Wykonanie histogram przedstawiający rozkład emisji CO2

```python
plt.figure(figsize=(8,5))
plt.hist(fuel["CO2EMISSIONS"].dropna(), bins=15, label="Emicja CO2")
plt.title("Histogram rozkładu emisji CO2")
plt.xlabel("Emisja CO2")
plt.ylabel("Wielkość emisji")
plt.legend()
plt.savefig("histogram.png")
plt.show()
```
