# Ćwiczenie 3
## Wykresy

### cena biletów dla pierwszych 15 rekordów

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.plot(titanic["Fare"].head(15),
    marker='o',
    linestyle='-',
    label="Cena biletu")
plt.title("Cena biletu pierwszych 15 pasażerów")
plt.xlabel("Numer pasażera")
plt.ylabel("Cena biletu")
plt.legend()
plt.savefig("wykres_liniowy_titanic.png")
plt.show()
```

### Wykres punktowy zależności wieku od ceny biletu

```python
plt.figure(figsize=(8,5))
plt.scatter(titanic["Age"], titanic["Fare"],
label="Pasażerowie")
plt.title("Wiek a cena biletu")
plt.xlabel("Wiek")
plt.ylabel("Cena biletu")
plt.legend()
plt.savefig("wykres_punktowy_titanic.png")
plt.show()
```

### Histogram wieku pasażerów

```python
plt.figure(figsize=(8,5))
plt.hist(titanic["Age"].dropna(), bins=15, label="Wiek")
plt.title("Histogram wieku pasażerów")
plt.xlabel("Wiek")
plt.ylabel("Liczba pasażerów")
plt.legend()
plt.savefig("histogram_titanic.png")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib as plt

df = titanic
```

### Utwórz nową cechę o nazwie „AgeGroup”, która będzie przydzielać każdego pasażera statku do jednej z czterech grup wiekowych.

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 18, 35, 60, 100],
    labels=["Dziecko", "Młody", "Dorosły", "Senior"]
)
```

### liczba osób w każdej grupie wiekowej

```python
df["AgeGroup"].value_counts()
```

### Utwórz cechę o nazwie „FamilySize”, która określa, ile członków rodziny podróżuje z danym pasażerem.


```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```
