```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

iris = pd.read_csv('../pliki/iris.csv')
iris.head()
```

```python
iris.info()
```

### wyznacz zmienne wejsciowe

```python
X = iris.iloc[:,0:4]
```

### standaryzacja danych

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### metoda PCA dla dwóch składowych

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

### udział wyjaśnionej wariancji

```python
pca.explained_variance_ratio_
```

### wykres dwóch pierwszych składowych

```python
plt.figure(figsize=(8,6))
plt.scatter(
 X_pca[:,0],
 X_pca[:,1]
)
plt.xlabel("Pierwsza składowa")
plt.ylabel("Druga składowa")
plt.title("PCA - Iris")
plt.grid(True)
plt.show()
```
