```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

df = pd.read_csv('../pliki/twitter-data.csv')
df.head()
```

### usuniecie wartosci typu null

```python
df.dropna()
df.isnull().sum()
```

```python
df.isna().mean().sort_values(ascending=False)
```

```python
df["category"] = df["category"].fillna(df["category"].median())
df.info()
```

```python
df = df.dropna(subset=["clean_text"])
df.info()
```

### tokenizacja

```python
text_column = df.columns[0]
tokens = df[text_column].apply(word_tokenize)
tokens.head()
```

### stemming

```python
stemmer = PorterStemmer()
stems = tokens.apply(lambda x: [stemmer.stem(word) for word in x])
stems.head()
```

### lematyzacja

```python
lemmatizer = WordNetLemmatizer()
lemmas = tokens.apply(lambda x: [lemmatizer.lemmatize(word)
for word in x])
lemmas.head()
```

### usuwanie słów stop wyświetlając efekt przeprowadzonego działania.

```python
stop_words = set(stopwords.words('english'))
filtered = tokens.apply(
 lambda x: [word for word in x if word.lower() not in stop_words]
)
filtered.head()
```
